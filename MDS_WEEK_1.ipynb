### Comparing Python List vs. NumPy Array Performance

```python
import time
import numpy as np

# Define the size of the array/list for performance comparison
size = 10000000 # 10 million elements

# --- Python List ---
# Create a Python list with a range of numbers
python_list = list(range(size))

# Record the start time for list sum calculation
start_time_list = time.time()
# Calculate the sum of elements in the Python list
sum_list = sum(python_list)
# Record the end time for list sum calculation
end_time_list = time.time()

# Calculate the time taken for list sum
time_taken_list = end_time_list - start_time_list
# Print the time taken and the sum for the Python list
print(f"Python List Sum Time: {time_taken_list:.6f} seconds")
print(f"Sum of Python List: {sum_list}")
```

```python
# --- NumPy Array ---
# Create a NumPy array with a range of numbers
numpy_array = np.arange(size)

# Record the start time for NumPy array sum calculation
start_time_numpy = time.time()
# Calculate the sum of elements in the NumPy array
sum_numpy = np.sum(numpy_array)
# Record the end time for NumPy array sum calculation
end_time_numpy = time.time()

# Calculate the time taken for NumPy array sum
time_taken_numpy = end_time_numpy - start_time_numpy
# Print the time taken and the sum for the NumPy array
print(f"NumPy Array Sum Time: {time_taken_numpy:.6f} seconds")
print(f"Sum of NumPy Array: {sum_numpy}")
```

### Comparing Python List vs. NumPy Array Memory Usage

```python
import sys
import numpy as np

# Define the size for memory usage comparison
size = 10000000

# Create a Python list for memory calculation
python_list_mem = list(range(size))

# Get the memory size of the Python list object itself (overhead)
list_overhead = sys.getsizeof(python_list_mem)

# Print memory usage details for Python list
print(f"Memory of Python List (object overhead only): {list_overhead} bytes")
print(f"Approximate memory for 10 million Python integer objects (each ~28 bytes): {size * 28} bytes")
print(f"Total estimated memory for Python List (including elements): {list_overhead + (size * 28)} bytes")
```

```python
# Create a NumPy array for memory calculation
numpy_array_mem = np.arange(size)

# Get the memory size of the NumPy array in bytes
numpy_memory = numpy_array_mem.nbytes

# Print memory usage for NumPy array
print(f"Memory of NumPy Array: {numpy_memory} bytes")
```
